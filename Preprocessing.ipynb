Stratified Sampling and Splitting the Data

```python
from sklearn.model_selection import train_test_split
```

```python
RANDOM_STATE=42
X_sample, _, y_sample, _ = train_test_split(
    calc_val, 
    observed_output,
    train_size=30000,
    stratify=observed_output,
    random_state=RANDOM_STATE)
print ("sample:", X_sample.shape, y_sample.value_counts(normalize=True).to_dict())
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_sample,
    y_sample,
    stratify = y_sample,
    test_size = 0.2,
    random_state=RANDOM_STATE)
print ("train:", X_train.shape, y_train.value_counts(normalize=True).to_dict())
print ("test:", X_test.shape, y_test.value_counts(normalize=True).to_dict())
```

Model Training

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.svm import SVC
```

```python
from sklearn.metrics import (
    accuracy_score, roc_auc_score, precision_score, recall_score,
    f1_score, matthews_corrcoef, confusion_matrix
)
```

For KNN and SVM using Mini subset

```python
X_tune, _, y_tune, _ = train_test_split(
    X_trainS,
    y_train,
    train_size = 5000,
    random_state = RANDOM_STATE,
    stratify = y_train
)
myData_ratio = y_tune.mean().round(3)
print("The subset dimension:", X_tune.shape)
print("The wildfire ratio:", myData_ratio)
```

Hyperparameter Test

```python
import time
```

```python
def tune_param(candidates, factory, X, y, label, cv=3):
    scores = {}
    for value in candidates:
        start_time = time.time()
        model_instance = factory(value)
        cv_accuracy = cross_val_score(
            model_instance, X, y, cv=cv, scoring="accuracy", n_jobs=-1
        ).mean()
        scores[value] = cv_accuracy
        elapsed_time = time.time() - start_time
        print(f"  Testing setting [{label}={value!r:>10}] -> CV Accuracy: {cv_accuracy:.4f} ({elapsed_time:.1f}s)")
    best_setting = max(scores, key=scores.get)
    best_score = scores[best_setting]
    print(f"  The best {label} is: {best_setting!r} (Best CV Accuracy: {best_score:.4f})\n")
    return best_setting
```

Logistic Regression

```python
from sklearn.model_selection import cross_val_score
```

```python
best_lr_model_val = tune_param(
    candidates=[0.01, 0.1, 1.0, 10.0],
    factory=lambda C: LogisticRegression(C=C, max_iter=2500, random_state=RANDOM_STATE),
    label="C",
    X=X_trainS,
    y=y_train,
)

lr_mod = LogisticRegression(
    C=best_lr_model_val, max_iter=2500, random_state=RANDOM_STATE
)
lr_mod.fit(X_trainS, y_train)
```

Decision Tree

```python
best_des_tree = tune_param(
    [3, 5, 7, 10, None],
    lambda d: DecisionTreeClassifier(max_depth=d, random_state=RANDOM_STATE),
    X_trainS,
    y_train,
    "max_depth",
)
dt_mod = DecisionTreeClassifier(max_depth=best_des_tree, random_state=RANDOM_STATE)
dt_mod.fit(X_trainS, y_train)
```

KNN

```python
best_KNN = tune_param(
    [3, 5, 7, 9, 11],
    lambda k: KNeighborsClassifier(n_neighbors=k),
    X_tune,
    y_tune,
    "n_neighbors",
)
knn_mod = KNeighborsClassifier(n_neighbors=best_KNN)
knn_mod.fit(X_trainS, y_train)
```

Gaussian Naive Bayes

```python
gauss_n_b_mod = GaussianNB()
gauss_n_b_mod.fit(X_trainS, y_train)
```

Support Vector Machine

```python
best_svm = tune_param(
    [0.1, 1.0, 10.0],
    lambda C: SVC(kernel = "rbf", C=C, probability=False, random_state = RANDOM_STATE),
    X_tune,
    y_tune,
    "C",
)
t0 = time.time()
svm_mod = SVC(kernel = "rbf", C=best_svm, probability=True, random_state = RANDOM_STATE)
svm_mod.fit(X_trainS, y_train)
print (f"Final fit:{time.time()-t0:.1f}s\n")
```

Model Evaluation

```python
models = {
    "Logistic Regression": lr_mod,
    "Decision Tree":       dt_mod,
    "KNN":                 knn_mod,
    "Gaussian NB":         gauss_n_b_mod,
    "SVM":                 svm_mod,
}

print("Trained models:", list(models.keys()))
```

```python
rows = []
conf_matrices = {}

for name, model in models.items():
    y_pred  = model.predict(X_testS)
    y_proba = model.predict_proba(X_testS)[:, 1]
    rows.append({
        "Model":     name,
        "Accuracy":  accuracy_score(y_test, y_pred),
        "AUC":       roc_auc_score(y_test, y_proba),
        "Precision": precision_score(y_test, y_pred),
        "Recall":    recall_score(y_test, y_pred),
        "F1":        f1_score(y_test, y_pred),
        "MCC":       matthews_corrcoef(y_test, y_pred),
    })
    conf_matrices[name] = confusion_matrix(y_test, y_pred)

results_df = pd.DataFrame(rows).set_index("Model").round(4)
print("Comparison table (test set, n=6000):")
print(results_df.to_string())

print("\nPer-model confusion matrices (rows = true, cols = predicted):")
for name, cm in conf_matrices.items():
    tn, fp, fn, tp = cm.ravel()
    print(f"\n{name}")
    print(f"             pred=0   pred=1")
    print(f"  true=0   {tn:>7}  {fp:>7}")
    print(f"  true=1   {fn:>7}  {tp:>7}")
```

Save the Models

```python
os.makedirs ("model", exist_ok = True)
model_files = {
    "Logistic Regression": "logistic_regression.joblib",
    "Decision Tree": "decision_tree.joblib",
    "KNN": "knn.joblib",
    "Gaussian NB": "naive_bayes.joblib",
    "SVM": "svm.joblib",
}
for name, mdl in models.items ():
    joblib.dump (mdl, os.path.join("model", model_files[name]))
    print ("saved:", name)
```
